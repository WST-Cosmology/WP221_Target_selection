```python
import pickle
def save_pickle(dat, filename, **kwargs):
    file = open(filename,'wb')
    pickle.dump(dat, file)
    file.close()
def load(filename, **kwargs):
    with open(filename, 'rb') as fin:
        return pickle.load(fin, **kwargs)
```

```python
path = '../survey_design/telescope_and_science_metrics/'
```

```python
survey_design_bright = load(path + 'survey_design_Bright_desilike_bao_rsd_forecast.pkl')
survey_design_grey = load(path + 'survey_design_Grey_desilike_bao_rsd_forecast.pkl')
survey_design_dark_qso = load(path + 'survey_design_Dark_qso_only_bao_rsd_fnl_forecast.pkl')
survey_design_grey_MagMax = load(path + 'survey_design_Grey_MagMax.pkl')
```

```python
n=0
for k, tracer in enumerate(['BG_bright', 'BG_grey', 'LRG','ELG', 'MagMax', 'QSO_qlf']):

    if tracer == 'BG_bright':
        survey = survey_design_bright
        i =7
        #plt.errorbar(survey_design_bright['per_tracer_forecasts'][tracer+'_list_zbin_rsd'][i], survey_design_bright['per_tracer_forecasts'][tracer+'_list_sigma_rsd'][i], xerr=0.1)
    if tracer == 'BG_grey':
        survey = survey_design_grey
        i=2

    if tracer == 'LRG':
        survey = survey_design_grey
        i=-1 #22.8

    if tracer == 'ELG':
        survey = survey_design_grey
        i=-2 #25.3

    if tracer == 'MagMax':
        survey = survey_design_grey_MagMax
        i=12
    if tracer == 'QSO_qlf':
        survey = survey_design_dark_qso
        i=18

    if tracer!='MagMax' and tracer!='QSO_qlf':
        n+=survey['config_survey'][tracer+'_target_density'][i]

    if tracer=='MagMax': 
        print(n)
    
    print(tracer, '=', survey['config_survey'][tracer+'_target_density'][i])


    mask_color = np.array(survey['config_survey']['tracers'])==tracer
   
    color = np.array(survey['config_survey']['color'])[mask_color][0]
    
    mag_Str = np.array(survey['config_survey']['limiting_mag_band'])[mask_color][0]
    #print(survey['config_survey'][tracer+'_spec_density'][i])

    max_mag = survey['config_survey'][tracer + '_mag_centers'][i]

    print(tracer, f' mag = {max_mag:.2F}')
    print()
    label = tracer + f' ({mag_Str} < {max_mag:.1f})'
    if tracer!='MagMax':
        x, y = survey['per_tracer_forecasts'][tracer+'_list_zbin_rsd'][i], 100*np.array(survey['per_tracer_forecasts'][tracer+'_list_sigma_rsd'][i])
        plt.errorbar(np.array(x)[y < 10], np.array(y)[y < 10], xerr=0.1, label = label,
                     color = color, 
                     marker='o', fmt='-')

    if tracer=='MagMax':
        x = np.array(
                survey['per_tracer_forecasts'][tracer + '_list_zbin_rsd'][i]
            )
        y = 100 * np.array(
                survey['per_tracer_forecasts'][tracer + '_list_sigma_rsd'][i]
            )
            
        mask = y < 10
        
        x_plot = x[mask]
        y_plot = y[mask]
        
        xerr = 0.1
        
        plt.errorbar(
            x_plot, y_plot,
            xerr=xerr,
            label=label,
            color=color,
            marker='o',
            #fmt='-'
        )
        
        plt.fill_betweenx(
            y_plot,
            x_plot - xerr,
            x_plot + xerr,
            color=color,
            alpha=0.4
        )

            
        #plt.plot(np.array(x)[y < 10], np.array(y)[y < 10],
                     #color = color, 
                     #marker='x',)
        

    plt.xlim(0, 3)
    plt.ylim(0, 11)
plt.ylabel(r'$\sigma(f\sigma_8) (\%)$', fontsize=14)
plt.xlabel(r'redshift bin ($\Delta z = 0.2)$', fontsize=14)
plt.tick_params(axis='both', labelsize=14)
plt.legend(fontsize=11, ncols=2)
plt.grid()

plt.savefig(f'../figures_for_paper/desilike_magmax_rsd_redshift_bins.png', dpi = 200, bbox_inches='tight',)
```
