```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from astropy.table import Table; 
from scipy.interpolate import RectBivariateSpline
```

```python
z_axis_interp = np.linspace(0, 4, 30)
m_max_axis_interp = np.linspace(21, 25, 20)
f = np.zeros([len(z_axis_interp), len(m_max_axis_interp)])

z_arr_file = np.linspace(0, 6, 100)
m_max_arr_file = [21, 22, 23, 23.3, 23.5, 24, 24.5, 25]
store_f = np.zeros([len(z_arr_file), len(m_max_arr_file)])

name_file = [21, 22, 23, '23p3',  '23p5', 24, '24p5', 25]

for i, rmax in enumerate(name_file):
    t = Table.read(f"qlf/qlf_rmax_{rmax}.ecsv", format="ascii.ecsv")
    
    if rmax == '23p3': 
        rmax_ = '23p5'
    else: rmax_ = rmax

    z = t['z']
    qlf =  t[f'qlf_{rmax_}']
    qlf_interp = np.interp(z_arr_file, z, qlf)
    arr = np.arange(0, 4.05, 0.05)
    ntot = np.sum(np.interp(arr, z, qlf))
    print(rmax, ntot)
    
    dn_dz = ntot * qlf_interp/np.trapz(qlf_interp,z_arr_file, )
    store_f[:,i] = dn_dz

from scipy.interpolate import RegularGridInterpolator

interp = RectBivariateSpline(
    z_arr_file,
    m_max_arr_file,
    store_f,
    kx=3,
    ky=3
)

Z, M = np.meshgrid(z_axis_interp, m_max_axis_interp, indexing='ij')
f = np.clip(interp.ev(Z, M), 0, 1000000)
```

```python
desi_qso_target = 310
desi_qso = 205
#desi_qso_target = desi_qso + desi_contamination
#desi_qso_target = alpha * qso + beta * desi_qso_target
#desi_qso_target(1 - beta) = alpha * qso
#desi_qso_target = alpha * qso / (1 - beta)
```

```python
eff = desi_qso/desi_qso_target
```

```python
eff
```

```python
np.interp(23, m_max_axis_interp, np.trapz(f, z_axis_interp, axis=0))
```

```python
alpha = desi_qso/np.interp(23, m_max_axis_interp, np.trapz(f, z_axis_interp, axis=0))
beta = 1 - desi_qso/desi_qso_target
```

```python
alpha
```

```python
beta
```

```python
ratio = alpha/(1-beta)
```

```python
#plt.figure(figsize=(6,5))

plt.plot(m_max_axis_interp, np.trapz(f, z_axis_interp, axis=0), '-k', label = 'Palanque-Delabrouille+2016', lw=1)
#plt.plot(m_max_axis_interp, np.trapz(f[z_axis_interp >= 2.1,:], z_axis_interp[z_axis_interp >= 2.1], axis=0), '-b', label = r'Palanque-Delabrouille+2016 $z>2.1$')

plt.plot(m_max_axis_interp, ratio*np.trapz(f, z_axis_interp, axis=0), ':r', label = 'targets (This work)')
plt.plot(m_max_axis_interp, eff*ratio*np.trapz(f, z_axis_interp, axis=0), '--g', label = 'confirmed QSOs (This work)')
plt.plot(m_max_axis_interp, eff*ratio*np.trapz(f[(z_axis_interp >=2.1),:], z_axis_interp[(z_axis_interp >=2.1)], axis=0), '--C1', label = r'confirmed QSOs $z > 2.1$ (This work)')
plt.plot(m_max_axis_interp, eff*ratio*np.trapz(f[(z_axis_interp >= 0.8)*(z_axis_interp <=3.1),:], z_axis_interp[(z_axis_interp >= 0.8)*(z_axis_interp <=3.1)], axis=0), '--m', label = r'confirmed QSOs $0.8 < z < 3.1$ (This work)')
plt.ylabel(r'QSO density [deg$^{-2}$]', fontsize=12)
plt.xlabel(r'$r_{\rm max}$', fontsize=14)
plt.tick_params(axis='both', labelsize=12)
plt.xlim(22, 25)
plt.ylim(0, 800)
#plt.vlines(23, -1, 1000, color = 'darkred', lw=0.5,  label = 'DESI DR1 magnitude cut', zorder=0)

plt.scatter(23, 310, marker='X', color = 'r', edgecolor='k', zorder=1000, s=80)
plt.scatter(23, 205,marker = 'X', color = 'g', edgecolor='k', zorder=1000, s=80)
plt.scatter(23, 60,color = 'C1',marker = 'X', edgecolor='k', zorder=1000, s=80)
plt.scatter(23, 186,color = 'm',marker = 'X', edgecolor='k', zorder=1000, s=50)
plt.scatter([], [], color='k', marker='X', edgecolor='k',  s=50, label = r'DESI DR1 QSO $r_{\rm max}=23$')

plt.grid()
plt.legend(framealpha=1, fontsize=10)
plt.savefig(f'../../../figures_for_paper/qso_versus_mmax.png', dpi = 200, bbox_inches='tight',)
```
